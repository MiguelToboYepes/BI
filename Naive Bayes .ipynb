## **Modelo de Preaprobacion**
* **Caso de estudio:** Se quiere implementar un modelo de preaprobacion de creditos a partir de las variables socioeconommicas
Estas variables son:
* Edad
*Ingresos
*Egresos
*Monto (EAD)


La variable de referencia sera la variabale Preaprobacion. Esta indica la pre aprobacion o pre negacion de un credito.


1. Se cargan las librerias de trabajo

```python
# %load_ext cudf.pandas
import pandas as pd
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix
```

2. Cargar datos de trabajo

```python
nxl='/content/sample_data/6. SolicitantesCrédito(USD) (1).xlsx'
XDB=pd.read_excel(nxl,sheet_name=0) #DataFrame : Tabla de datos - Tablas de Word

#Seleccionamos las variables de trabajo
XDB=XDB[['Edad','Ingresos','Egresos','Monto (EAD)','PreApr']]
filas=np.where(~np.isnan(XDB['PreApr']))[0]
XDB=XDB.iloc[filas,:]
XDB=XDB.dropna()


#Separar variables de entrada - variables de salida
XD=XDB[['Edad','Ingresos','Egresos','Monto (EAD)']]
yd=XDB['PreApr']#Variables de Salida
display(XDB.head())
```

3. **Entrenar el modelo Naive Bayes**

```python
# Suppress scientific notation
np.set_printoptions(suppress=True)
pd.set_option('display.float_format', lambda x: '%.3f' % x)
```

```python
#Crear el Modelo
mnb=GaussianNB()

#Entrenar el Modelo
mnb.fit(XD,yd)

# Datos por categoria
ndat=mnb.class_count_
print('Los datos por categoria son:\n',ndat)
cat=mnb.classes_
print('Las categorias son:\n',cat)

#Determinamos las propiedad estadisticas de las variables
u=mnb.theta_ #La media de variables por categoria
print('La Media de variables es:\n:',u)
sigma=np.sqrt(mnb.var_) #La desviacion estandar de variables por categoria
print('La desviacion estandar de variables es:\n:',sigma)
print('Los limites superiores de las variables son:\n',u+sigma)
print('Los limites inferiores de las variables son:\n',u-sigma)

```

4. Se procede con la evaluación de desempeño del modelo( frente al pronostico de la decision)

Falsos: Equivocacion

Verdaderos: Acierto



```python
ydp=mnb.predict(XD)
cm=confusion_matrix(yd,ydp)
print('La matriz de confusion es:\n',cm)
```

5. Metricas de desempeño

```python
cm=confusion_matrix(yd,ydp)
print('La matriz de confusion es:\n',cm)
VN=cm[0,0]
FP=cm[0,1]
FN=cm[1,0]
VP=cm[1,1]
TD=len(XD) #Total datos

#Exactitud: Muestra el comportamiento general del modelo
Ex=(VP+VN)/TD
print('La exactitud del modelo es:\n',Ex)

#Tasa de error:Porcentaje de equivocacion
ter=(FP+FN)/TD
print('La tasa de error del modelo es:\n',ter)

#Sensibilidad: Como se comporta el modelo frente a los pre aprobados
Se=VP/(VP+FN)
print('La sensibilidad del modelo es:\n',Se)

#Especificidad: Como se comporta el modelo frente a los pre negados
Esp=VN/(VN+FP)
print('La especificidad del modelo es:\n',Esp)

#Precisión: Capacidad del modelo para predecir VP
Pre=VP/(VP+FP)
print('La precision del modelo es:\n',Pre)

#Precision Negativa: Capacidad del modelo para predecir VN
PreN=VN/(VN+FN)
print('La precision negativa del modelo es:\n',PreN)
```

# **Analisis de Resultados**
* De acuerdo con la base de datos hay un total de 5482 datos, los cuales estan
divididos en 2959 en pre negados y 2883 en pre aprobados.

* Se puede observar igualmente que la categoria de pre aprobados presento los valores mas altos frente a las variable Edad, Ingresos, Egresos, Monto.

* De la matriz de confusion se puede observar que el modelo tiende a equivocarse en mayor medida en el pronostico de los valores pre apobados con un total de 947 falsos negativos, frente  a 457 de Falsos Positivos

* Con las metricas de desempeño basadas en la matriz de confusion, podemos observar que la exactitud del modelo es del 76% lo que muestra el buen comportamiento del modelo frente al pronostico de los datos en general,la tasa de error del modelo es de 24%

* Estas metricas tambien nos muestran que la sensibilidad del modelo a pre aprobar creditos es relativamente baja 67%, mientras la especificidad para negarlos es de 84% por lo tanto se comporta mejor diagnosticando los negativos

* En la precision podemos ver que es mas preciso a pre aprobarlos con una precision del 81%, mientras la precision al momento de pre negarlos es del 73%
